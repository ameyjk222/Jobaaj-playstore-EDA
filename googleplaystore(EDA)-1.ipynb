```python
### How many apps have a rating of 4.0 and above ?

import pandas as pd
df = pd.read_excel("playstore_data.xlsx")
df

high_rated_apps = df[df["Rating"] >= 4.0]
print("high_rated_apps:", len(high_rated_apps))
```

```python
### What is the average number of reviews for free vs paid apps ?

df["Reviews"]= pd.to_numeric(df["Reviews"], errors="coerce")
```

```python
avg_reviews = df.groupby("Type")["Reviews"].mean()

print(avg_reviews)
```

```python
### What is the average app size for each category ?

def convert_size(size):
    if size == "Varies with device":
        return None
    elif "M" in size:
        return float(size.replace("M", ""))
    elif "K" in size:
        return float(size.replace("k", "")) / 1024
    else:
        return None

df["Size_MB"]= df["Size"].apply(convert_size)
```

```python
avg_size = df.groupby("Category")["Size_MB"].mean()

print(avg_size)
```

```python
### How many apps were last updated in 2018 ?

df["Last Updated"] = pd.to_datetime(df["Last Updated"], errors ="coerce")

Updated_apps_2018 = df[df["Last Updated"].dt.year == 2018]

print("Updated_apps_2018:", len(Updated_apps_2018))
```

```python
### What is the correlation between the number of installs and the app rating ?

import pandas as pd
import numpy as np

df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")

correlation = df["Installs"].corr(df["Rating"])

print(correlation)
```

```python
### Which app categories have the highest average rating ?

print(df["Category"].dtype)
print(df["Rating"].dtype)

avg_ratings = df.groupby("Category")["Rating"].mean()

print(avg_ratings)
```

```python
### How does the price of an app affect its average rating ?
 
print(df["Price"].dtype)

paid_apps = df[df["Type"] == "Paid"].copy()

paid_apps["Price"] = paid_apps["Price"].astype(str).str.replace("$", "", regex="False")
paid_apps["Price"] = pd.to_numeric(paid_apps["Price"], errors="coerce")

paid_apps["Rating"] = pd.to_numeric(paid_apps["Rating"], errors="coerce")

avg_price_by_rating = df.groupby("Price")["Rating"].mean()

print(avg_price_by_rating)
```

```python
### What is the distribution of app ratings across different content ratings ?

import seaborn as sns

df["Rating"]= pd.to_numeric(df["Rating"], errors="coerce")

plt.Figure(figsize=(10,6))
sns.boxplot(x="Content Rating", y="Rating", data=df)

plt.title("Distribution of app by Content Rating")
plt.xlabel("Content Rating")
plt.ylabel("Rating")
plt.show()
```

```python
### Which genres have the most apps with over 1 million installs ?

print(df["Installs"].dtype)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")


highly_install_apps = df.groupby("Installs")["Genres"].count()
highly_install_apps = df[df["Installs"] > 1000000]

genre_count = (highly_install_apps.groupby("Genres").size().sort_values(ascending=False))
print(genre_count)
```

```python
### How frequently do apps get updated ? Calculate the average time between updates.

df["Last Updates"] = pd.to_datetime(df["Last Updated"], errors="coerce")

df = df.sort_values("Last Updated")

average_time = df["Last Updates"].mean()

print(average_time)
```

```python
### What is the impact of app size on the number of installs ?

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

def convert_size(size):
    if size == "Varies with device":
        return None
    elif "M" in str(size):
        return float (str(size).replace("M", ""))
    elif "k" in str(size):
        return float (str(size).replace("k", "")) / 1024
    else:
        return None

df["Size_MB"] = df["Size"].apply(convert_size)




df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

plt.Figure(figsize=(10, 6))
sns.scatterplot(x="Installs", y="Size_MB", data=df)
plt.title("Installs by app sizes")
plt.xlabel("Sizes")
plt.ylabel("No. of Installs")
plt.show()
```

```python
### Which apps have the highest number of reviews, and what are their ratings ?

df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")

top_reviews = (df.sort_values(by="Reviews", ascending=False)
               [["App", "Reviews", "Rating"]]
               .head(10)
            )

print(top_reviews)
```

```python
### How does the content rating distribution differ between free and paid apps ?

content_rating_distribution = pd.crosstab(
    df["Content Rating"],
    df["Type"]

)

print(content_rating_distribution)
```

```python
### What are the top 5 categories with the most installs ?

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

top_categories = df.groupby("Category")["Installs"].sum().sort_values(ascending=False).head(5)

print(top_categories)
```
